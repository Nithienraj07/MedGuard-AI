# MedGuard AI - Exploratory Data Analysis
This notebook covers the Medical Appointment No Shows dataset.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Ensure raw data directory exists
os.makedirs('../data/raw', exist_ok=True)
csv_path = '../data/raw/KaggleV2-May-2016.csv'

if not os.path.exists(csv_path):
    print('Dataset not found at', csv_path)
    print('Please download the dataset from Kaggle:')
    print('URL: https://www.kaggle.com/datasets/joniarroba/noshowappointments')
    print('Extract the downloaded archive and place KaggleV2-May-2016.csv in the data/raw/ folder.')
else:
    print('Dataset found! Ready to proceed.')
```

```python
if os.path.exists(csv_path):
    df = pd.read_csv(csv_path)
    display(df.head())
else:
    print('Please download the dataset first.')
```

```python
if os.path.exists(csv_path):
    df.info()
```

```python
if os.path.exists(csv_path):
    display(df.describe())
```

```python
if os.path.exists(csv_path):
    display(df.isnull().sum())
```

```python
if os.path.exists(csv_path):
    sns.countplot(x='No-show', data=df)
    plt.title('Distribution of Target Variable (No-show)')
    plt.show()
```
